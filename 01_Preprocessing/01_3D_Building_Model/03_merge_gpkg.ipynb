This script merges all individual GeoPackage files into a single GeoPackage containing all LoD2 building footprints.

```python
import geopandas as gpd
import pandas as pd
from pathlib import Path

# Directory containing the individual GeoPackages
input_dir = Path(r"C:\Users\agz90fk\Documents\Masterarbeit\03_Daten\Input\LoD2\download\individual_gpkgs")

# Output GeoPackage
output_gpkg = Path(r"C:\Users\agz90fk\Documents\Masterarbeit\03_Daten\Output\Analysis_w_LoD2\LoD2\LoD2_2025.gpkg")

# Collect all GeoPackage files
gpkg_files = list(input_dir.glob("*.gpkg"))

gdfs = []

for gpkg in gpkg_files:
    print(f"Reading {gpkg.name}")
    gdf = gpd.read_file(gpkg)
    gdfs.append(gdf)

# Merge all GeoDataFrames
merged_gdf = gpd.GeoDataFrame(
    pd.concat(gdfs, ignore_index=True),
    crs=gdfs[0].crs
)

# Write the merged GeoPackage
merged_gdf.to_file(output_gpkg, driver="GPKG")

print("✅ Done!")
```
