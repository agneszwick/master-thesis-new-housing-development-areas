Merge the results of LST and NDVI Comparison.

```python
import geopandas as gpd

# =============================================================================
# CONFIGURATION
# =============================================================================

ndvi_file = r"C:\Users\agz90fk\Documents\EO4CAM\3_Daten\Output\Masterarbeit\Comparison_NHDA_RA_v2\NDVI_Comparison\NDVI_analysis_geometries.gpkg"
lst_file = r"C:\Users\agz90fk\Documents\EO4CAM\3_Daten\Output\Masterarbeit\Comparison_NHDA_RA_v2\LST_Comparison\LST_analysis_geometries.gpkg"

ndvi_layer = "analysis_areas"
lst_layer = "analysis_areas"

# Output
output_file = r"C:\Users\agz90fk\Documents\EO4CAM\3_Daten\Output\Masterarbeit\Comparison_NHDA_RA_v2\Comparison_LST_NDVI.gpkg"
output_layer = "merged"

# Join columns
join_cols = ["nhda_id", "type"]

# =============================================================================
# LOAD DATA
# =============================================================================

ndvi = gpd.read_file(ndvi_file, layer=ndvi_layer)
lst = gpd.read_file(lst_file, layer=lst_layer)

print(ndvi.columns)
print(lst.columns)

# Remove unnecessary duplicate columns
lst = lst.drop(columns=["geometry", "area_ha", "n_years"])
ndvi = ndvi.drop(columns=["n_years"])

# =============================================================================
# MERGE (JOIN)
# =============================================================================

merged = ndvi.merge(
    lst,
    on=join_cols,
    how="inner"
)

# =============================================================================
# SAVE OUTPUT
# =============================================================================

merged.to_file(output_file, layer=output_layer, driver="GPKG")

print("✅ Done! File saved to:", output_file)
```
